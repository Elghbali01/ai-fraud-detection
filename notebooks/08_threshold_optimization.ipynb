# 08 — Validation threshold optimization

The threshold is selected by maximum F2 on the 42,722-row validation partition, weighting recall more than precision. This is a development objective, not an asserted business cost. The final-test threshold is fixed before the test metrics are computed.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
results = json.loads((ROOT / "reports/ml/latest_results.json").read_text(encoding="utf-8"))
threshold = results["threshold"]
print("Policy:", threshold["policy"])
print("Selected threshold:", threshold["threshold"])
validation_point = pd.Series(
    threshold["validation_metrics_at_selected_threshold"],
    name="Validation operating point",
)
display(validation_point.to_frame())
curve = pd.read_csv(ROOT / "reports/ml/threshold_analysis.csv")
nearest = (curve["threshold"] - threshold["threshold"]).abs().nsmallest(8).index
display(curve.loc[nearest].sort_values("threshold"))
display(Image(filename=str(ROOT / "reports/ml/figures/threshold_analysis.png")))
```
