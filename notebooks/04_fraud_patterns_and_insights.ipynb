# 04 — Fraud patterns and insights

## Objective
Inspect descriptive differences between labeled classes without overstating what anonymized features establish.

## Analytical questions
- What amount summaries differ by historical label?
- Which numeric features have the strongest rank association with the label?
- How stable are time-bin comparisons when bins have few transactions?

## Methodology
Use class-conditional summaries and Spearman coefficients. Do not interpret outliers as confirmed fraud, use feature labels as business concepts, or infer causation.

## Code and results

```python
from fraud_detection.analytics import (
    amount_summary,
    class_feature_associations,
    class_imbalance,
    time_bin_summary,
)
from fraud_detection.notebook_support import load_notebook_frame

frame, data_status = load_notebook_frame()
print(f"Data source status: {data_status}")
if data_status == "synthetic_fixture":
    print("All observations below are generated fixture values and not ULB evidence.")
elif data_status == "unverified_local_file":
    print("The local file passed schema validation; its source provenance is unverified.")
print("Label counts:", class_imbalance(frame)["counts"])
print("Class-conditional amount summaries:")
display(amount_summary(frame))
print("Strongest descriptive feature-label associations:")
display(class_feature_associations(frame).head(10))
print("Elapsed-time bins; low_volume indicates the configured count floor was not reached:")
display(time_bin_summary(frame, minimum_transactions=1_000))
```

## Interpretation
Observed differences can guide later hypotheses but do not show that transaction amount or a PCA component causes fraud. Mission 02 should use time-aware validation as well as stratified baselines, then report uncertainty and threshold trade-offs.
