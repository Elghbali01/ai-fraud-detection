# 06 — Supervised model benchmark

Candidates share three expanding-window folds inside the chronological training partition. Average Precision is the primary selection measure. The post-freeze validation columns use an uncalibrated 0.5 reference threshold for description only; they did not change the model winner or selected operating threshold.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
results = json.loads((ROOT / "reports/ml/latest_results.json").read_text(encoding="utf-8"))
comparison = pd.read_csv(ROOT / "reports/ml/model_comparison.csv")
columns = [
    "candidate",
    "mean_average_precision",
    "std_average_precision",
    "mean_roc_auc",
    "validation_average_precision_at_0_5",
    "validation_roc_auc_at_0_5",
    "validation_precision_at_0_5",
    "validation_recall_at_0_5",
    "validation_false_positive_at_0_5",
    "validation_false_negative_at_0_5",
    "cv_fit_seconds",
    "validation_fit_seconds",
]
display(comparison.sort_values("mean_average_precision", ascending=False)[columns])
print("Selected by the frozen training-fold rule:", results["selected_model"]["name"])
print("One-standard-error rule:", results["selection"])
display(Image(filename=str(ROOT / "reports/ml/figures/model_comparison.png")))
```
